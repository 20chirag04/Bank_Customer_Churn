## Importing Liberaries

```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (OneHotEncoder,
                                OrdinalEncoder,
                                StandardScaler,)
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
import pickle
```

### Importing and Reading dataset 

```python
raw = pd.read_csv('Customer-Churn-Records.csv')
df = raw.copy()
df.head()
```

### Dropping Unnecessary Features

```python
cols_to_drop = ['RowNumber','CustomerId','Surname','Complain']
df.drop(columns=cols_to_drop,inplace=True)
```

```python
df.head()
```

```python
num_cols = ['CreditScore','Age','Balance','EstimatedSalary','Point Earned']
cat_cols = ['Geography','Gender','Tenure','Card Type']
ordinal_cols = ['Satisfaction Score']
```

### Split the Data into Train and Test

```python
X = df.drop('Exited',axis = 1)
y = df['Exited']
```

```python
X.shape,y.shape
```

```python
X_train,X_test,y_train,y_test = train_test_split(
    X,y,test_size=0.25,random_state=12
)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

### Columns Transformation

```python
num_pipeline = Pipeline(
    steps = [
        ('scaler',StandardScaler())
    ]
)
cat_pipeline = Pipeline(
    steps = [
        ('OneHotEncoding', OneHotEncoder(drop='first'))
    ]
)
ordinal_pipeline = Pipeline(
    steps = [
        ('OrdinalEncoding',OrdinalEncoder())
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers= [
        ('Numerical',num_pipeline,num_cols),
        ('Categorical',cat_pipeline,cat_cols),
        ('Ordinal',ordinal_pipeline,ordinal_cols)
    ],
    remainder="passthrough"
)
```

```python
X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

### Saving the Preprocessor File

```python
with open('preprocessor.pkl','wb') as file:
    pickle.dump(preprocessor,file)
```
