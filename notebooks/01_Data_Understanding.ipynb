# Bank Custumer Churn
As we know, it is much more expensive to sign in a new client than keeping an existing one.
It is advantageous for banks to know what leads a client towards the decision to leave the company.
Churn prevention allows companies to develop loyalty programs and retention campaigns to keep as many customers as possible.

In business, churn means the rate at which customers stop using a company's product or service or employees leave their jobs over a specific time period. It is commonly tracked as a percentage using metrics like customer churn or revenue churn.

### Importing and Reading Dataset

```python
raw_data = pd.read_csv("Customer-Churn-Records.csv")
```

```python
df = raw_data.copy()
```

```python
df.head()
```

- RowNumber: corresponds to the record (row) number and has no effect on the output.
- CustomerId : contains random values and has no effect on customer leaving the bank. 
- Surname : the surname of a customer has no impact on their decision to leave the bank. 
- CreditScore : can have an effect on customer churn, since a customer with a higher credit score is less likely to leave the bank. 
- Geography : a customer’s location can affect their decision to leave the bank. 
- Gender : gender of customers 
- Age : Age of customers
- Tenure : refers to the number of years that the customer has been a client of the bank.
- Balance : balance in their bank accounts. 
- NumOfProducts : refers to the number of products that a customer has purchased through the bank. 
- HasCrCard : denotes whether or not a customer has a credit card.
- IsActiveMember : active customers or not. 
- EstimatedSalary : Monthly customer salaries. 
- Exited : whether or not the customer left the bank. 
- Complain : customer has complaint or not.
- Satisfaction Score : Score provided by the customer for their complaint resolution. 
- Card Type : type of card hold by the customer. Points Earned—the points earned by the customer for using credit card.

### Understanding columns types 

```python
num_cols = [feature for feature in df.columns if df[feature].dtype != "object" ]
print(num_cols)
```

```python
category_cols = [feature for feature in df.columns if df[feature].dtype == "object" ]
print(category_cols)
```

```python
discrete_num_cols = [feature for feature in df[num_cols] if df[feature].nunique() < 15]
print(discrete_num_cols)
```

```python
low_cardinality_cols = [feature for feature in df[category_cols] if df[feature].nunique() < 15]
print(low_cardinality_cols)
```

#### Value Counts

```python
for i in df[discrete_num_cols]:
    print(df[i].value_counts())
    print("-"*25)
```

```python
for i in df[low_cardinality_cols]:
    print(df[i].value_counts())
    print("-"*25)
```

## Data Learnings
- Satisfaction Score is an ordinal categorical feature. 
- Card Type is low cardinality ordinal Data.
- Data is Mostly clean have no outliers and null values.
- RowNumber, CustomerId, and Surname can be removed as they do not provide useful predictive information.
- low cardinality columns can be converted using label encoding and OneHotEncoding.
- Target Feature will be "Exited" ,where:
  - 0 : Customer Stayed
  - 1 : Customer Churned
