# Exploratory Data Analysis

### Importing Neccessary Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Importing and Reading Data

```python
df = pd.read_csv("Customer-Churn-Records.csv")
df.head()
```

## Univariate Analysis

### Categorical Features

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(df['Geography'])
plt.subplot(1,2,2)
sns.histplot(df['Gender'])
plt.show()
```

```python
df['Satisfaction Score'].value_counts()
```

### Numerical Features

```python
plt.figure(figsize=(15,6))
plt.subplot(1,3,1)
sns.histplot(df['CreditScore'],kde=True)
plt.subplot(1,3,2)
sns.histplot(df['Balance'])
plt.subplot(1,3,3)
sns.histplot(df['Age'])
plt.show()
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(df['Point Earned'])
plt.subplot(1,2,2)
sns.histplot(df['Tenure'])
plt.show()
```

## Bivariate Analysis

### Numerical Features vs Target

```python
## Looking at the data set again
df.head()
```

```python
sns.histplot(data= df,x= 'Age',hue='Exited')
```

```python
plt.figure(figsize=(6,4))
sns.histplot(data= df, x= 'Balance', hue='Exited')
```

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.histplot(data=df, x='CreditScore',hue='Exited')
plt.subplot(1,2,2)
sns.histplot(data=df,x= 'EstimatedSalary',hue='Exited')
plt.show()
```

### Categorical Features

```python
plt.figure(figsize=(15,6))
plt.subplot(1,2,1)
sns.countplot(data=df,x='Geography',hue='Exited')
plt.subplot(1,2,2)
sns.countplot(data=df,x='Gender',hue='Exited')
plt.show()
```

```python
sns.barplot(data=df,x='Card Type',y='EstimatedSalary',)
```

```python
sns.countplot(data=df,x='Card Type',hue='Exited')
```

```python
sns.violinplot(x="Exited", y="Age", data=df)
```

```python
sns.countplot(data=df,x='HasCrCard',hue='Exited')
```

## Multivariate Analysis

```python
df.corr(numeric_only=True)
```

```python
pd.pivot_table(
    df,
    values="Exited",
    index="Geography",
    columns="Gender",
    aggfunc="mean"
)
```

```python
pd.crosstab(
    df["Gender"],
    df["Exited"],
    normalize="index"
) * 100
```

```python
pd.crosstab(
    df["Geography"],
    df["Exited"],
    normalize="index"
) * 100
```

```python
df.head()
```

```python
df["Exited"].value_counts()
```

```python
df.groupby('Exited')[['CreditScore','Balance','EstimatedSalary','Age']].mean()
```

```python
df.groupby(['Gender','Geography'])['Exited'].mean()
```

# EDA Learnings
- The target variable 'Exited' is imbalanced, with more customers staying than churning.
- France have most number of customers.
- feMale customer population have higher churn rate.  
- Card Type does not show a strong relationship with customer churn.
- No significant outliers were observed.
